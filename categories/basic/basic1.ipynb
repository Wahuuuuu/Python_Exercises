**Given a list[int], return the max without using max()**

```python
# test 5 times
for i in range(5):
    # init list
    l = []
    for j in range(5):
        j1 = random.randint(-5, 5)
        l.append(j1)

    # traverse for max
    max = -10086
    for num in l:
        if num > max: max = num

    print(l, ", max = ", max)
```

```python
for i in range(5):
    l = [random.randint(-5, 5) for j in range(5)]

    max = l[0]
    for j in l[1:]:
        if j > max: max = j

    print(f"{l}, max = {max}")
```

**Given a list, returns the most frequent element**

```python
# Init the problem
elems = [random.randint(1, 3) for _ in range(10)]

# Solve the problem
elem_dict = {}
for elem in elems:
    elem_dict[elem] = elem_dict.get(elem, 0) + 1

max_elem = -1
max_elem_times = -1
for elem, times in elem_dict.items():
    if times > max_elem_times: 
        max_elem = elem
        max_elem_times = times

print(f"list: {elems}\n dict: {elem_dict}\n most freq elem is {max_elem}, appeared {max_elem_times} times.")
```

**Realize def compress(s)**

```python
def compress(s: string) -> string:
    compressed = []

    prev_elem = ""
    for elem in s:
        if prev_elem != elem:
            compressed.append(elem)
            compressed.append(1)
            prev_elem = elem
        else:
            compressed[-1] += 1

    return "".join(str(i) for i in compressed)
```
