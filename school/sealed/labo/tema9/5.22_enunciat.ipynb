### 5.22 Ordenar llista aparellada

Suposem que necessitem ordenar llistes els nombres de les quals apareixen duplicats, un darrere l'altre. Per exemple, la llista [1, 1, 4, 4, 3, 3, 5, 5]. Prova l' algorisme d'ordenació mergesort. Calcula la complexitat d'aquest algorisme a nivell de comparacions i d'intercanvi d'elements.

No pots utilitzar funcions d'ordenació de Python.

El mergesort ha de retornar la llista modificada.

```python
pair: int = 2

def mergesort_parelles(llista: list[int]) -> list[int]:
    """
    Aquesta funció va partint la llista en una banda dreta i una esquerra
    fins que aquestes bandes tenen un element. 
    Quan ho aconsegueix, crida merge per fusionar ambdues bandes de manera 
    ordenada fins arribar a fusionar la llista completa original.

    Idea:
        The elements of the list are strictly pair formed
        even index are the first number of the pair, 
        odd index are the second number of the pair
    """
    l: int = len(llista)
    if l <= pair: 
        return llista

    points_right = lambda l: (l//2) if (l%4 == 0) else (l//2 + 1)
    mid = points_right(l)
    # mid points the fist element of the first pair of the right part

    left_part = mergesort_parelles(llista[0 : mid])
    right_part = mergesort_parelles(llista[mid:])

    return (merge(left_part, right_part)) 
```

```python
def merge(left_part: list[int], right_part: list[int]) -> list[int]:
    aux: list[int] = []
    i = 0; lim_i = len(left_part)
    j = 0; lim_j = len(right_part)

    while i < lim_i and j < lim_j:
        if left_part[i] <= right_part[j]:
            aux.extend([left_part[i], left_part[i]])
            i += pair
        else:
            aux.extend([right_part[j], right_part[j]])
            j += pair
    
    aux.extend(left_part[i : lim_i])
    aux.extend(right_part[j : lim_j])
    return aux
```

```python
llista = [1, 1, 7, 7, 2, 2, 3, 3, 5, 5]
assert mergesort_parelles(llista) == [1, 1, 2, 2, 3, 3, 5, 5, 7, 7]
```
