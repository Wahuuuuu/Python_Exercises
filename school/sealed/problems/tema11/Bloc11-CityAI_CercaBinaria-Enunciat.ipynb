## 🧠 Objectius

Aprofundir el l'algoritme de cerca binària i treballar-lo per adaptar-lo a casos concrets.

Per a facilitar la realització dels exercicis es proveeix el codi de l'algorisme de cerca binària.

```python
# Codi original de la funció iterativa de cerca binària amb finestres

def cerca_binaria(A:list[any],K:any):
    inici = 0
    final = len(A)-1
    while inici <= final:
       mig = (inici + final) // 2
       if A[mig] > K:
            final = mig - 1
       elif A[mig] < K: 
            inici = mig + 1
       else: 
           return mig                           
    return -1
```

## ✍️ 🔎➗ Exercici 1: Menor i major relatius

L'algorisme ha de tenir un cost sublineal.

Fes servir aquesta plantilla per escriure el teu codi:

```python
import math
```

```python
# Plantilla per a l'exercici.

def menor_major_relatius(nums: list[int], k: int) -> (int, int):
    """
    Aquesta funció troba el el valor immediatament més petit que k 
    i el valor immediatament més gran que k. 
    En qualsevol dels dos casos, retorna -1 si aquest valor no existeix.
    
    Parameters
    ----------
    nums: list[int]
    k: int
    
    Returns
    -------
    menor: int
    major: int
    """ 
    
    index_menor = searching_menor(nums, k)
    index_major = searching_major(nums, k)

    value_or_minus_one = lambda index, nums: nums[index] if 0 <= index < len(nums) else -1

    menor = value_or_minus_one(index_menor, nums)
    major = value_or_minus_one(index_major, nums)
    return menor, major
```

```python
def searching_menor(l: list[int], value: int) -> int:
    """
    This function returns the index of the last number in l that number < value
    If there isn't any number in l that number < value, the index is -1

    Parameter
    ---
    l: list[int] in increasingly sorted
    value: int

    Return
    ---
    index_menor: int, the index of the last number in l that number < value
    """
   

    """
    Invariant:
    - for all i <= left, l[i] < value is true
    - for all i > right, l[i] >= value is true
    - the the interval [left, right] must contains the index of target

    Once the loop terminates, left == right is true
    """
    left = -1
    right = len(l) - 1
    while (left < right) :
        i = math.ceil((left + right) / 2)  # i will never points to left

        if l[i] < value:
            left = i
        else: # l[i] >= value
            right = i - 1
    
    return left
```

```python
def searching_major(l: list[int], value: int) -> int:
    """
    This function returns the index of the index of the first number that number > value
    If the number doesn't exists, return len(l)

    Parameter
    ---
    index_menor: int, index of the last number that number < value
    l: list[int] increasingly sorted
    value: int

    Return
    ---
    index_mayor: int, theindex of the first number in l that number > value
    """

    """
    Invariant:
    - for all i < left, l[i] <= value
    - for all i >= right, l[i] > value
    - the index must be in the interval [left, right]

    Note:
    Once the loop terminates, left == right is true
    """
    left = 0
    right = len(l)
    while left < right:
        i = (left + right) // 2  # the i never points to right

        if l[i] <= value:
            left = i + 1
        else: # l[i] > value
            right = i
    
    return right
```

```python
assert menor_major_relatius([1,4,6,8,9],3) == (1,4)
assert menor_major_relatius([1,4,6,8,9], 4) == (1,6)
assert menor_major_relatius([1,4,6,8,9], 5) == (4,6)
assert menor_major_relatius([1,4,6,8,9],10) == (9,-1)
assert menor_major_relatius([5],3) == (-1,5)
```

## ✍️ 🔎➗ Exercici 2: Troba la primera i darrera ocurrència d'un nombre donat en una llista ordenada

Donada una llista d'enters i un nombre, escriu un algorisme que retorni la posició de la primera i darrera ocurrència d'aquest nombre a la llista. Si el nombre no apareix, el programa ha de retornar -1, -1 com a índexs.

```python

# Plantilla per a l'exercici.

def primera_darrera_ocurrencia(llista: list[int], nombre: int) -> (int, int):
    """
    Aquesta funció troba la primera i darrera ocurrència d'un valor a la llista
     
    Parameters
    ----------
    llista: list[int]
    nombre: int
    
    Returns
    -------
    pos_primera_ocurrencia: int
    pos_darrera_ocurrencia: int
    """

    index_last_smaller = searching_menor(llista, nombre)
    index_first_bigger = searching_major(llista, nombre)

    """
    if llista[smaller + 1] != nombre, nombre not in llista, -1, -1
    if llista[smaller + 1] == nombre, smaller + 1 es la primera ocurrencia
        bigger - 1 = ultim ocurrencia
    """
    pos_primera_ocurrencia = index_last_smaller + 1
    if pos_primera_ocurrencia >= len(llista) or llista[pos_primera_ocurrencia] != nombre:
        # nombre not in llista
        return -1, -1
    
    pos_darrera_ocurrencia = index_first_bigger - 1
    return pos_primera_ocurrencia, pos_darrera_ocurrencia
```

## ✍️ 🔎➗ Exercici 3: Existeix

Donada una llista $L$ de $n$ nombres enters ordenats de manera estrictament creixent (i.e. no hi ha repeticions), crea una funció "existeix", que, usant l'estratègia de dividir i vèncer, comprovi si existeix algun element de la llista que compleixi el següent:

$$L[i] = i, i ∈ \{0, ..., n-1\}$$

L'algorisme proposat ha de tenir complexitat $O(log(n))$.

Fes servir aquesta plantilla per escriure el teu codi:

```python
# Plantilla per a l'exercici.

def existeix(llista: list[int]) -> bool:
    """
    Aquesta funció comprova si existeix un element tal que llista[i] = i.
    
    Parameters
    ----------
    llista: list[int]
    
    Returns
    -------
    b_existeix: bool
    """

    """
    Invariant:
    - for all i < left, we have l[i] < i
    - for all i > right, we have l[i] > i
    - if exists an i == l[i], i must belongs to llista[left : right+1]
    """
    left, right = 0, len(llista) - 1
    while (left <= right):
        mid = (left + right) // 2

        value = llista[mid]
        if value < mid:
            left = mid + 1
        elif value > mid:
            right = mid - 1
        else:
            return True
        
    b_existeix = False
    return b_existeix
```

```python
assert existeix([1, 4, 5, 6, 7]) == False
assert existeix([0, 1, 2, 7]) == True
assert existeix([-1, 1, 4, 8]) == True
assert existeix([-5, -3, 0, 1, 5, 7]) == False
```
