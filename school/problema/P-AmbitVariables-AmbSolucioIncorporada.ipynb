## ✍️ 1. 📘 Problemes amb l'àmbit de les variables

En aquest notebook molts dels codis es mostren perquè l'objectiu és principalment demostratiu, i veure el codi és essencial per entendre el resultat.

**1**. Escriu una funció que defineixi una variable local i intenta accedir-hi des de l'exterior de la funció.

```python

def mostrar_variable_local() -> None:
    variable_local: int = 42
    print("Dins de la funció:", variable_local)


mostrar_variable_local()
# Aquesta línia provoca NameError: la variable només existeix dins la funció.
print(variable_local)
```

**2**. Defineix una variable global i una funció amb una variable local que té el mateix nom. Observa quina variable es fa servir dins la funció.

```python

variable_global: int = 10


def mostrar_variables() -> None:
    variable_local: int = 5
    print("Dins de la funció:", variable_local)
    print("Dins de la funció, accedint a la global:", variable_global)


mostrar_variables()
print("Fora de la funció, accedint a la global:", variable_global)
```

**3**. Defineix una funció amb un paràmetre i intenta accedir-hi des de l'exterior de la funció.

```python

def mostrar_parametre(parametre: int) -> None:
    print("Valor del paràmetre dins de la funció:", parametre)


mostrar_parametre(42)
# Aquesta línia provoca NameError: el paràmetre només existeix dins la funció.
print(parametre)
```

**4**. Escriu una funció que modifiqui el valor d'una variable global.

```python

variable_global: int = 10


def modificar_variable_global() -> None:
    global variable_global
    variable_global += 5


print("Valor global abans de la funció:", variable_global)
modificar_variable_global()
print("Valor global després de la funció:", variable_global)
```

**5**. Crea una funció exterior i una funció interior. Intenta accedir a la variable local de la funció interior des de l'exterior.

```python

def mostrar_variable_interior() -> None:
    def mostrar_variable_local() -> None:
        variable_interior: int = 5
        print("Dins de la funció interior:", variable_interior)

    mostrar_variable_local()
    # Aquesta línia provoca NameError: la variable és local a la funció interior.
    print("Dins de la funció exterior:", variable_interior)


mostrar_variable_interior()
```

**6**. Defineix una variable global i una funció amb una variable local que té el mateix nom. Observa quina variable es fa servir quan la funció s'invoca des de la funció interior.

```python

variable_global: int = 10


def mostrar_variable_local() -> None:
    variable_local: int = 5
    print("Dins de la funció:", variable_local)


print(variable_global)
mostrar_variable_local()
print("Però des de fora:", variable_global)
```

**7**. Defineix una variable global i una funció que rep un argument amb el mateix nom. Observa quina variable es fa servir dins de la funció.

```python

variable_global: int = 10


def mostrar_argument(variable_global: int) -> None:
    print("Dins de la funció:", variable_global)


mostrar_argument(5)
print("Fora de la funció:", variable_global)
```

**8**. Crea una funció que afegirà un element a una llista global.

```python

llista_global: list[int] = [1, 2, 3]


def afegir_a_llista(element: int) -> None:
    llista_global.append(element)


afegir_a_llista(4)
print(llista_global)
```

**9**. Defineix una llista global i una funció que crea una llista local amb el mateix nom. Comprova quina llista s'utilitza dins de la funció.

```python

llista_global: list[int] = [1, 2, 3]


def mostrar_llista_local() -> None:
    llista_local: list[int] = [4, 5, 6]
    print("Dins de la funció:", llista_local)


mostrar_llista_local()
print("Fora de la funció:", llista_global)
```

**10**. Intenta modificar una variable global des de dins d'una funció sense declarar-la com a global.

```python

variable_global: int = 10


def modificar_sense_global() -> None:
    # Aquesta assignació provoca UnboundLocalError, no NameError:
    # Python considera variable_global local abans de poder-la llegir.
    variable_global += 5


modificar_sense_global()
print(variable_global)
```

**11**. Defineix una funció que utilitzi una variable local dins d'un bucle.

```python

def mostrar_resultats_del_bucle() -> None:
    for comptador in range(5):
        resultat: int = comptador * 2
        print("Dins del bucle:", resultat)


mostrar_resultats_del_bucle()
```

**12**. Defineix una funció que utilitzi una variable global dins d'un bucle.

```python

variable_global: int = 0


def acumular_valors() -> None:
    global variable_global
    for comptador in range(5):
        variable_global += comptador
        print("Dins del bucle:", variable_global)


acumular_valors()
```
