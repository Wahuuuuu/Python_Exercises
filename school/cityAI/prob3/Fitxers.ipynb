## Fitxers

Sovint les dades que utilitza un algorisme es troben emmagatzemades de
manera persistent en fitxers.

En aquesta assignatura treballarem principalment amb **fitxers de text**.

## Rutes dels fitxers

Quan indiquem només un nom de fitxer:

```python
open("prova.txt")
```

Python interpreta aquesta ruta respecte del **directori de treball actual**.

En un notebook Jupyter, aquest directori sovint coincideix amb la carpeta
del notebook, però **no és una garantia**. Per tant, és més precís parlar
del directori de treball que de «la carpeta on hi ha el notebook».

Podem consultar-lo amb:

```python
from pathlib import Path

print(Path.cwd())
```

## Obrir fitxers

La funció `open()` permet obrir un fitxer.

Alguns modes habituals són:

| Mode | Significat |
|---|---|
| `"r"` | lectura |
| `"w"` | escriptura; crea el fitxer o substitueix el contingut existent |
| `"a"` | afegeix contingut al final |
| `"x"` | crea un fitxer nou i falla si ja existeix |

Per exemple:

```python
fitxer = open("prova.txt", "r", encoding="utf-8")
```

obre `prova.txt` en mode lectura.

> El mode `"r"` és una cadena i, per tant, ha d'anar entre cometes.

Quan treballem amb fitxers de text és convenient indicar explícitament
la codificació:

```python
encoding="utf-8"
```

Això evita que el comportament depengui de la configuració de codificació
del sistema operatiu.

## Gestió de fitxers amb `with`

Tot i que podem obrir i tancar manualment un fitxer:

```python
fitxer = open("prova.txt", "r", encoding="utf-8")
contingut = fitxer.read()
fitxer.close()
```

la forma recomanada és utilitzar un **gestor de context** amb `with`:

```python
with open("prova.txt", "r", encoding="utf-8") as fitxer:
    contingut = fitxer.read()
```

Quan acaba el bloc `with`, Python tanca automàticament el fitxer, fins i
tot si durant el processament es produeix una excepció.

Per això utilitzarem habitualment aquesta sintaxi.

## Escriure en un fitxer

```python
with open(
    "fitxerprova.txt",
    "w",
    encoding="utf-8",
) as fitxer:
    fitxer.write("Hola món\n")
    fitxer.write("Aquest és el nostre fitxer de text nou\n")
    fitxer.write("I això és una línia nova\n")
    fitxer.write("El fitxer s'ha creat.\n")
```

El caràcter `\n` representa un **salt de línia**.

El contingut del fitxer serà:

```text
Hola món
Aquest és el nostre fitxer de text nou
I això és una línia nova
El fitxer s'ha creat.
```

> **Atenció:** obrir un fitxer existent amb mode `"w"` n'elimina el
> contingut anterior.

## Llegir tot el contingut

El mètode `read()` retorna el contingut del fitxer com una cadena (`str`):

```python
with open(
    "fitxerprova.txt",
    "r",
    encoding="utf-8",
) as fitxer:
    contingut = fitxer.read()

print(contingut)
```

## Llegir totes les línies

`readlines()` retorna una llista de cadenes:

```python
with open(
    "fitxerprova.txt",
    "r",
    encoding="utf-8",
) as fitxer:
    linies = fitxer.readlines()

print(linies)
```

El resultat pot ser semblant a:

```python
[
    "Hola món\n",
    "Aquest és el nostre fitxer de text nou\n",
    "I això és una línia nova\n",
    "El fitxer s'ha creat.\n",
]
```

Els caràcters `\n` formen part de les cadenes llegides.

## Recórrer un fitxer línia a línia

Un fitxer és iterable, de manera que el podem recórrer directament:

```python
with open(
    "fitxerprova.txt",
    "r",
    encoding="utf-8",
) as fitxer:
    for linia in fitxer:
        print(linia, end="")
```

Utilitzem `end=""` perquè cada línia llegida ja conté habitualment el seu
propi `\n`. Si utilitzéssim simplement:

```python
print(linia)
```

`print()` hi afegiria un segon salt de línia i apareixerien línies en blanc
entre les línies del fitxer.

## Conversió de dades llegides

Quan llegim dades d'un fitxer de text, obtenim **cadenes**.

Per exemple, si una línia conté:

```text
25
```

la lectura ens retorna `"25"`, no l'enter `25`.

Si necessitem un nombre, l'hem de convertir:

```python
edat = int(text)
```

De la mateixa manera, si volem interpretar un valor booleà, és preferible
definir explícitament quins textos acceptem com a certs o falsos en lloc
d'utilitzar `bool(text)`.

## ✍️ 1. Exercici 1: esport, joc o aplicació preferida

Fes una funció que llegeixi informació sobre l'esport, joc o aplicació preferida de l'usuari i n'escrigui el resultat. La funció ha de:

1. llegir les dades des d'un fitxer de text;
2. convertir les dades al tipus adequat quan sigui necessari;
3. escriure el resultat en un altre fitxer de text.

Les dades han d'incloure com a mínim:

- un text;
- un nombre;
- un valor booleà.

Utilitza `with open(...)` tant per llegir com per escriure els fitxers i
indica `encoding="utf-8"`.

````python
def process_info(info: str) -> str:
    """ dado una data original, procesar la data y retornarla
    Parameters
        data: str
            con el formato "<nombre del juego> <tiempo por dia> <si es de multijugadores>\n"
    
    ------
    Return: str
        "Le gusta <nombre del juego>, juega <minuts por dia>, (no) es un juego de multijugadores"
    """
    data = info.strip().split()
    minuts = int(data[1]) * 60
    info_multijugadores = "es un juego de multijugadores" if data[2] == "True" else "no es un juego de multijugadores"

    return "".join([data[0], " ", str(minuts), " ", info_multijugadores, "\n"])


with open("desti_joc.txt", "w", encoding = "utf-8") as destination:
    with open("info_joc.txt", "r", encoding = "utf-8") as source:
        for line in source:
            original_info = line
            processed_info = process_info(original_info)
            destination.write(processed_info)

````
