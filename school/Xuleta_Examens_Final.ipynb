## Elements bàsics

*Programa* Un programa en Python és una seqüència de de definicions i comandes.

*Comanda* Una comanda és una instrucció directa a l'interpret per fer alguna cosa. El procés per determinar el resultat d'una comanda es diu avaluació.

*Literal* Un literal és una entitat el valor de la qual és ella mateixa i que per tant no cal avaluar. 3 és un literal.

*Objectes* Els objectes són les entitats bàsiques que manipula Python. Tots els objectes tenen un _tipus_ que defineix quines operacions podem aplicar-lis. Els tipus poden ser escalars o no escalars. Els primers són indivisibles, els segons tenen estructura interna.

*Tipus escalars* Python té varios tipus escalars:
    
+ int S'usa per representar els nombres enters: 3, 1245 o -23.
+ float S'usa per representar els nombres reals: 3.0, 1245.2325 o -23.56.
+ bool S'usa per representar valors els valors Booleans True o False.
+ None És un tipus amb un únic valor.

*Operadors* Els operadors ens permeten combinar objectes, formant _expressions_, cada una de les quals denota un objecte, anomenat _valor_, d'un determinat tipus. L'expressió 3 * 2 denota l'objecte 5 de tipus int. Els operadors dels tipus int i float són:
    
+ i+j S'usa per representar la suma. Si tant i com j son int, el resultat és int. Si algun dels dos és float, el resultat es float.
+ i-j S'usa per representar la resta. Si tant i com j son int, el resultat és int. Si algun dels dos és float, el resultat es float.
+ i*j S'usa per representar el producte. Si tant i com j son int, el resultat és int. Si algun dels dos és float, el resultat és float.
+ i//j S'usa per representar la divisió entera (retorna el quocient i s'oblida del reste): el valor de 6//2 és 3 i el valor de 6//4 és 1.
+ i/j S'usa per representar la divisió.
+ i%j S'usa per representar la resta (o mòdul) de la divisió entera de i per j.
+ i**j S'usa per representar $i^j$ . Si tant i com j són int, el resultat és int. Si algun dels dos és float, el resultat és float.
+ &gt;,<,>=,<=, ==, != Són els operadors de comparació, el resultat dels quals és de tipus bool.

Tots aquests operadors segueixen l'ordre de precedència habitual.

Els operadors del tipus bool són:

+ a and b Si tant a com b són True, el resultat és True. Si algun dels dos és False, el resultat és False.
+ a or b Si a o b són True, el resultat és True. Si els dos són False, el resultat és False.
+ not a Si a és True, el resultat és False. Si a és False, el resultat és True.

## Condicionals

Fins ara hem vist seqüencies linials d'instruccions, que són executades per l'interpret Python una darrera l'altra. Si volem un esquema d'execució condicional, en arbre, necessitem especificar tres parts:

+ 1. Un test (que es una expressió de tipus bool).
+ 2. Un bloc de codi que s'executi quan l'expressió prengui el valor True.
+ 3. Un bloc de codi, opcional, que s'executi quan l'expressió prengui el valor False.

Les instruccions que implementen aquest esquema s'anomenen condicionals i tenen aquesta forma:

```python
a:int = 10

if a > 3.0:
    b:float = 0.0
else:
    b:bool = True
```

Python usa l'estructura visual del codi (definida per les identacions de
cada línia) com a part de la seva semàntica. Concretament, el codi que
s'executa quan l'expressió booleana pren un determinat valor ha
d'estar en un nivell superior d'identació que el test. L'identació de
Python són 4 caràcters en blanc.

Aquesta regla es pot aplicar a múltiples nivells:

```python
x:int = 5

if x%2 == 0:
    if x%3 == 0 :
        print('Divisible per 2 i per 3')
    else:
        print('Divisible per 2 i no per 3')
elif x%3 == 0:
    print('Divisible per 3 pero no per 2')
else:
    print('No es divisible ni per 2 ni per 3')
```

## Càlcul numèric

La manipulació d'objectes de tipus float i l'aplicació repetida de
càlculs han de ser tractats amb compte. Considerem aquest codi:

```python
x:float = 0.0
for i in range(10):
    x = x + 0.1
print(x == 1.0)
```

El resultat de la seva execució és False. Per entendre el motiu, cal
entendre com es representen els nombres binaris.

*Representació binària d'un nombre enter* Un nombre binari es
representa per una seqüència de dígits que poden ser 0 o 1,
anomenats bits. El bit de més a la dreta representa el valor 20,
el següent 21 i així successivament. Per tant, el nombre binari
101 correspon a 1 * 4 + 0 * 2 + 1 * 1 = 5.

*Representació binària d'un nombre real* Tots els llenguatges de
programació representen els nombres reals amb la representació
anomenada punt flotant, que correspon al producte dels dígits
significatius del nombre per un exponent. En el cas binari tant
els dígits significatius com l'exponent s'escriuen en forma
binària. El problema és que hi ha nombres que no es poden
representar de forma exacta (també passa en base 10), com
1 / 10 . A Python, hi ha 53 bits de precisió, i per tant el nombre
0.1 seria: 11001100110011001100110011001100110011001100110011001,
que equival al nombre:
0.1000000000000000055511151232157827021181583404541015625.

Quan en el programa anterior comparavem x amb 1.0, el
resultat era fals perque la suma binaria de les diferents
representacions de 0.1 no sumen 1.0.

Aquest fet és important quan fem tests d'igualtat. Sempre que
tinguem valors flotants, enlloc de fer x == y és millor usar
aquesta expressió d'igualtat (aproximada): abs(x-y) <
0.000001.

## Funcions

En teoria, amb els nombres, assignacions, input, comparacions i
iteracions, podem implementar ja qualsevol programa. Però no és
pràctic. Una de les possibilitats addicionals que proporciona qualsevol
llenguatge de programació és la d'agrupar instruccions en funcions.

*Definició* A Python, una funció es defineix de la següent manera:

```python
def max(x:int, y:int)->int:
    if x > y:
        return x
    else:
        return y
```

```python
max(4,5)
```

_def_ és una paraula reservada que indica que anem a definir una
funció. Seguidament escrivim el nom que volem donar a la
funció. La seqüència de noms que apareixen entre parèntesi
després del nom de la funció són els paràmetres, que prenen
valors en el moment que cridem la funció: max(3+2,4).

Seguidament ve el cos de la funció degudament identat. Si
volem que la funció retorni valors, hem de posar-hi la instrucció
return i l'expressió que volem que retorni.

*Execucio* Què passa quam cridem una funció? Primer, s'avaluen les
expressions corresponents als paràmetres. Després, el punt
d'execucióo de l'intèrpret es dirigeix a la primera línia de la
funció i s'executa el codi de la funció fins que trobem un return
o no queden instruccions per executar-se. En aquest darrer cas
la funció retorna el valor None. Per últim, el punt d'execució
torna a la instrucció posterior a la crida de la funció.

*Arguments* A Python hi ha dues maneres de passar arguments:
posicionalment o mitjançant paraules clau. Suposem aquesta
funció:

```python
def printName(firstName:str, lastName:str, reverse:bool):
    if reverse:
        print(lastName + ',' + firstName)
    else:
        print(firstName, lastName)
```

```python
printName('Donald', 'Knuth', True)
```

```python
printName('Donald', reverse = False, lastName = 'Knuth')
```

Podem cridar la funció posicionalment,
printName('Donald', 'Knuth', True),
i l'intèrpret assignarà valors als paràmetres en funció de l'ordre,
o mitjançant paraules clau,
printName('Donald', reverse = False, lastName = 'Knuth'),
on usa els noms.

*Scope* Aquest concepte correspon a l'espai de noms que pot veure
una funció. En principi, una funció pot veure els noms associats
als seus paràmetres i els de les variables definides en el seu
interior. Tots aquests només exisiteixen quan la funció s'està
executant.

A Python, les funcions també són objectes i per tant es poden
tractar com qualsevol altre objecte: tenen tipus, poden
apareixer en expressions, poden ser elements de llistes o es
poden passar com a paràmetres d'una altra funció.

## Codi dels algorismes principals estudiats durant el curs

```python
def fibonacci(n:int)->int:
    a, b = 0, 1
    for i in range (1, n+1):
        a,b = b, a+b
    return a

# L'algorisme de fibonacci optimitzat té una complexitat d'ordre O(n)
```

```python
def gcd(a:int,b:int)->int:
    while a:
        a, b = b%a, a
    return b

# L'algorisme de gcd té una complexitat d'ordre O(log (a))
```

```python
def mul(x:int,y:int)->int:
    import math
    if y == 0:
        return 0
    z:int = mul (x, math.floor(y/2))
    if y%2 == 0:
        return 2*z
    else:
        return x+2*z

# La multiplicació té una complexitat d'ordre O(n^2)
```

```python
import random

def fermat(num:int, test_count:int) -> bool:
    if num == 1:
        return False
    for x in range(test_count):
        val:int = random.randint(1, num-1)      # cost logaritmic respecte num bits 
        if pow(val, num-1, num) != 1:           # (log n)^3 on n és él num
            return False                   
    return True

# L'algorisme de Fermat té una complexitat d'ordre O(test_count · (log n)³), on n és el num
```

```python
def era(n)->set[int]:
    era1:set[int] = set((range(2,n)))  # tots els nombres 
    era2:set[int] = set([i for q in range(2,int(n**0.5)+1) 
                for i in range(q*2,n,q)]) # tots els múltiples
    return era1 - era2

# L'algorisme d'Eratóstenes té una complexitat d'ordre O(n log log n)
# El conjunt resultant tindrà O(sqrt(n) log n) elements
# Si volem convertir el resultat en llista podem fer list() amb cost O(sqrt(n) log n)
```

```python
def fibonacci(n: int) -> int:
    """
    Calcula el n-èssim nombre de Fibonacci 

    :param n: Índex (n ≥ 0) del nombre de Fibonacci a calcular.
    :return: El nombre de Fibonacci a la posició n.
    """
    a: int = 0
    b: int = 1

    for i in range(1, n + 1):
        a, b = b, a + b

    return a
```

```python
def BoyerMooreHorspool(patro:str, text:str)->int:
    m:int = len(patro)
    n:int = len(text)
    if m > n: 
        return -1
    taula_desplacaments = []                             # creem la taula buida
    for k in range(256):                                 # caràcters ASCII 
        taula_desplacaments.append(m)                    # per defecte el salt serà la llargada del patró
    for k in range(m - 1):                            
        taula_desplacaments[ord(patro[k])] = m - k - 1  #  per als caràcters del patró el salt és la distancia al final
                                                          # si apareixen més d'un cop es guarda el darrer només
    taula_desplacaments = tuple(taula_desplacaments)      # convertim en tupla perquè no ho tocarem 
    k = m - 1                                             # k és la posició del text que comparem amb la lletra final
                                                          # inicialment equival a la darrera lletra del patró
    while k < n:
        j = m - 1                                         # índex de la lletra del patró que comparem 
        i = k                                             # índex de la lletra del text que comparem
        while j >= 0 and text[i] == patro[j]:           
            j -= 1 
            i -= 1
                                                           # avancem(!) mentre hi hagi text i les lletres coincideixen
        if j == -1:                                        # si el patró s'ha acabat (totes les lletres han coincidit)  
            return i + 1                                   # retornem la posició del text
        k += taula_desplacaments[ord(text[k])]             # desplacem sense tenir en compte els avenços fets
    return -1    
```

```python
def distancia_levenshtein(patro, text):
    if len(text) == 0: 
        return len(patro)
    llargada_patro = len(patro) + 1
    llargada_text = len(text) + 1
    matriu_distancies = [[0] * llargada_text for x in range(llargada_patro)]
    for i in range(llargada_patro): 
        matriu_distancies[i][0] = i
    for j in range(llargada_text): 
        matriu_distancies[0][j] = j
    for i in range(1, llargada_patro):
        for j in range(1, llargada_text):
            deletion = matriu_distancies[i-1][j] + 1
            insercio = matriu_distancies[i][j-1] + 1
            substitucio = matriu_distancies[i-1][j-1]
            if patro[i-1] != text[j-1]: 
                substitucio += 1
            matriu_distancies[i][j] = min(insercio,deletion,substitucio)
    return matriu_distancies[llargada_patro-1][llargada_text-1]
```

```python
def subcadena_mes_llarga(cadena:str)->str:
    """
    Aquesta funció identifica la subcadena més llarga
    sense cap caràcter repetit.
    Parameters
    ----------
    cadena: string
    Cadena donada
    Returns
    -------
    subcadena: string
    Subcadena més llarga sense caràcters repetits
    """
    n:int = len(cadena)
    # Fem un diccionari amb els caràcters.
    diccCaracters:dict[str:str] = {caracter: False for caracter in cadena}
    # La solució es basa en una finestra que es redefineix cada cop que troba un caràcter repetit.
    iniciFinestra:int = 0
    # Inici i final de la finestra. El final és el darrer caràcter vist; 
    # el principi és el primer caràcter no repetit des del final de la finestra.
    iniciSubcadena:int = 0
    finalSubcadena:int = 0
    # Inici i final de la subcadena més llarga trobada fins al moment
    for finalFinestra in range(0, n):
        if diccCaracters[cadena[finalFinestra]]: # Si el caracter ja hi era.
            while (cadena[iniciFinestra]!= cadena[finalFinestra]):
                diccCaracters[cadena[iniciFinestra]] = False
                iniciFinestra += 1
                # Desplacem la finestra a partir de la primera aparició 
                # del caràcter, sense incloure'l.
            iniciFinestra += 1
        else:
            diccCaracters[cadena[finalFinestra]] = True
            # Anotem el caràcter com a existent.
            finalFinestra += 1
            # Com que no hi ha repeticions augmentem la finestra.
            if finalSubcadena - iniciSubcadena < finalFinestra - iniciFinestra:
                # Revisem que la nova subcadena no sigui més llarga que la que teníem guardada.
                iniciSubcadena = iniciFinestra
                finalSubcadena = finalFinestra
    return cadena[iniciSubcadena:finalSubcadena]
```

## 🎒 Algorisme de la motxilla

```python
from itertools import combinations
def motxilla_fb(nombre:int, capacitat:int, pes_cost:list[any]) -> (int,list[int]):
    """
    :param nombre: nombre d'items possibles
    :param capacitat: capacitat de la motxilla
    :param pes_cost: llista de tuples del tipus [(pes, cost), (pes, cost), ...]
    """
    millor_cost:int = None
    millor_combinacio:list[any] = []
    # genera totes les combinacions possibles d'1 o més elements
    for num_elements in range(nombre):
        for comb in combinations(pes_cost, num_elements + 1):
            pes:int = sum([pc[0] for pc in comb])
            cost:int = sum([pc[1] for pc in comb])
            if (millor_cost is None or millor_cost < cost) and pes <= capacitat:
                millor_cost = cost
                millor_combinacio = [0] * nombre
                for pc in comb:
                    millor_combinacio[pes_cost.index(pc)] = 1
    return millor_cost, millor_combinacio
```

## Johnson Trotter

```python
def johnson_trotter(n:int)->list[list[int]]:
    """
    Genera totes les N! permutacions del conjunt {1, 2, ..., n}.
    """
    # 1. Inicialització
    p:list[int] = list(range(1, n + 1))  
    direccions:list[bool] = [False] * n  # Direcció: False = esquerra (<-); True = dreta (->)
    permutacions:list[list[int]] = [list(p)] 
    hi_ha_moviment:bool = True

    # 2. Bucle de Generació
    while hi_ha_moviment:

        # A. Trobar l'element Mòbil més gran
        element_mobil:int = -1
        index_mobil:int = -1
        trobat_mobil:bool = False
        
        for i in range(n):
            k = p[i]         
            adj_index:int = i - 1 if direccions[i] == False else i + 1
            if 0 <= adj_index < n and k > p[adj_index] and k > element_mobil:
                element_mobil = k
                index_mobil = i
                trobat_mobil = True
        hi_ha_moviment = trobat_mobil

        # B. Intercanvi: Moure l'element mòbil
        target_index:int = index_mobil - 1 if direccions[index_mobil] == False else index_mobil + 1          
        p[index_mobil], p[target_index] = p[target_index], p[index_mobil]
        direccions[index_mobil], direccions[target_index] = direccions[target_index], direccions[index_mobil]
        
        # C. Canvi de Direcció: Revertir les direccions dels elements més grans       
        for i in range(n):
            if p[i] > element_mobil:
                direccions[i] = not direccions[i]
                
        # Afegir la nova permutació
        permutacions.append(list(p))

    return permutacions
```

## ♟️ Algoritme de les N Reines

```python
from itertools import permutations

def hi_ha_atac_diagonal(reines_per_columna:list[list[int]])->bool:
    """
    Comprova si dues reines s'ataquen diagonalment.
    'reines_per_columna' és una tupla (r0, r1, r2, ...), on ri és la fila de la reina a la columna i.
    """
    N:int = len(reines_per_columna)
    
    # Comprova totes les parelles de reines (i, j)
    for i in range(N):
        for j in range(i + 1, N):
            # i i j són les columnes
            # reines_per_columna[i] i reines_per_columna[j] són les files
            
            # Condició d'atac diagonal: |fila_i - fila_j| == |columna_i - columna_j|
            if abs(reines_per_columna[i] - reines_per_columna[j]) == abs(i - j):
                return True # Atac trobat
    return False

def n_reines_forca_bruta(N:int)->list[list[int]]:
    # Genera totes les permutacions de files (0 a N-1). 
    # Això garanteix una reina per fila i una per columna.
    files:list[int] = range(N)
    totes_disposicions:list[list[int]] = permutations(files)
    solucions:int = 0

    # Iterem sobre totes les disposicions (FORÇA BRUTA)
    for disposicio_files in totes_disposicions:
        # Si NO hi ha atac diagonal, és una solució vàlida
        if not hi_ha_atac_diagonal(disposicio_files):
            solucions += 1
            print(f"Solució trobada: {disposicio_files}") # Descomentar per veure les solucions

    return solucions
```

## Mergesort i la seva funció auxiliar, merge

```python
def merge(x:list[any],y:list[any])->list[any]:
    """
    Fusiona dues llistes prèviament ordenades en una nova llista ordenada
    :param
    x: una llista ordenada
    y: una llista ordenada
    :return
    una nova llista amb tots els elements d'x i y, ordenats
    """
    resultat:list[any] = []
    i:int = 0
    j:int = 0
    while(i < len(x) and j < len(y)):
        if (x[i] <= y[j]):
            resultat.append(x[i])
            i = i + 1
        else:
            resultat.append(y[j])
            j = j + 1
    resultat += x[i:]
    resultat += y[j:]
    return resultat
```

```python
def mergesort(l:list[any])->list[any]:
    """
    Ordena una llista
    :param
    l1: una llista a ordenar
    :return
    una nova llista ordenada
    """
    # print("MERGESORT: m'han cridat amb",l)  # xivato
    if len(l) < 2:
        return l
    else:
        meitat = len(l) // 2
        esquerra = mergesort(l[:meitat])
        dreta = mergesort(l[meitat:])
        return merge(esquerra, dreta)
```

## Quicksort

```python
def quick_sort(A:list[any]):
    quick_sort_r(A, 0, len(A) - 1)    
                                      
def quick_sort_r(A:list[any], primer:int, darrer:int):  
    if darrer > primer:
        pivot:int = partition(A, primer, darrer) 
        quick_sort_r(A, primer, pivot - 1) 
        quick_sort_r(A, pivot + 1, darrer)  
```

```python
def partition(A:list[any], primer:int, darrer:int):
    # primer intentem trobar un bon pivot
    meitat:int = (primer + darrer)//2  
    if A[primer] > A [meitat]: A[primer], A[meitat] = A[meitat], A[primer]
    if A[primer] > A [darrer]: A[primer], A[darrer] = A[darrer], A[primer]
    if A[meitat] > A[darrer]:   A[meitat], A[darrer] = A[darrer], A[meitat]  
    A[meitat], A[primer] = A[primer], A[meitat] 
    # inicialitzem, el pivot ha d'estar a la primera posició de la llista
    pivot:int = primer
    i = primer + 1
    j = darrer
    indexs_creuats:bool = False

    while not(indexs_creuats): 
        # anem avançant
        while i <= darrer and A[i] <= A[pivot]: i += 1
        while j >= primer and A[j] > A[pivot]: j -= 1
        if i >= j: 
            indexs_creuats = True
        else: 
            A[i], A[j] = A[j], A[i] 
    A[j], A[pivot] = A[pivot], A[j] 
    return j
```

## Cerca lineal

```python
def cerca_lineal(A:list[any],K:any):
  i:int=0
  while i < len(A) and A[i] != K:
    i += 1
  if i < len(A):
     return i
  else:
    return -1
```

## Cerca binària

```python
def cerca_binaria(A:list[any],K:any):
    inici = 0
    final = len(A)-1
    while inici <= final:
       mig = (inici + final) // 2
       if A[mig] > K:
            final = mig - 1
       elif A[mig] < K: 
            inici = mig + 1
       else: 
           return mig                           
    return -1
```
