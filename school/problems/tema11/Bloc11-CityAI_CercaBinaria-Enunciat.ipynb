## 🧠 Objectius

Aprofundir el l'algoritme de cerca binària i treballar-lo per adaptar-lo a casos concrets.

Per a facilitar la realització dels exercicis es proveeix el codi de l'algorisme de cerca binària.

```python
# Codi original de la funció iterativa de cerca binària amb finestres

def cerca_binaria(A:list[any],K:any):
    inici = 0
    final = len(A)-1
    while inici <= final:
       mig = (inici + final) // 2
       if A[mig] > K:
            final = mig - 1
       elif A[mig] < K: 
            inici = mig + 1
       else: 
           return mig                           
    return -1
```
