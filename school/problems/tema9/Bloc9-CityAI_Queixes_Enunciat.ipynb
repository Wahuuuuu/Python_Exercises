## 🧩 Context

L'Ajuntament de Barcelona recull les queixes dels ciutadans al llarg de l'any i les classifica pel seu tipus, per l'àrea de l'ajuntament que les ha de resoldre, per la data i pel lloc on s'han produit.

```python
import pandas as pd

path = "2025_IRIS_Peticions_Ciutadanes_OpenData.csv"
df = pd.read_csv(path)
df.head()

peticions = []

for _, row in df.iterrows():
    peticions.append({
        "id": row["FITXA_ID"],
        "tipus": str(row["TIPUS"]).replace('"', ''),
        "area": str(row["AREA"]).replace('"', ''),
        "detall": str(row["DETALL"]).replace('"', ''),
        "dia": int(row["DIA_DATA_ALTA"]),
        "mes": int(row["MES_DATA_ALTA"]),
        "any": int(row["ANY_DATA_ALTA"]),
        "districte": str(row["DISTRICTE"]).replace('"', ''),
        "barri": str(row["BARRI"]).replace('"', '')
    })

len(peticions)
```

## ✍️ Exercici 1: Adaptació de mergesort

Completa el codi del mergesort amb una variació: usarem una funció lambda "key" com a criteri d'ordenació.

A continuació us copiem el codi original i el codi adaptat on falten algunes parts.

```python
# Codi original de la funció merge

def merge(x:list[any],y:list[any])->list[any]:
    """
    Fusiona dues llistes prèviament ordenades en una nova llista ordenada
    :param
    x: una llista ordenada
    y: una llista ordenada
    :return
    una nova llista amb tots els elements d'x i y, ordenats
    """
    # print("MERGE: m'han cridat amb",x,y)  # xivato
    if len(x) < 1:
        return y
    if len(y) < 1:
        return x
    llista_nova = []
    i = j = 0
    while i < len(x) and j < len(y):
        if x[i] <= y[j]:
            llista_nova.append(x[i])
            i += 1
        else:
            llista_nova.append(y[j])
            j += 1
    llista_nova.extend(x[i:])
    llista_nova.extend(y[j:])
    return llista_nova
```

```python
# Codi original de la funció mergesort

def mergesort(l:list[any])->list[any]:
    """
    Ordena una llista
    :param
    l1: una llista a ordenar
    :return
    una nova llista ordenada
    """
    # print("MERGESORT: m'han cridat amb",l)  # xivato
    if len(l) < 2:
        return l
    else:
        meitat = len(l) // 2
        esquerra = mergesort(l[:meitat])
        dreta = mergesort(l[meitat:])
        return merge(esquerra, dreta)
```

```python
# Codi adaptat de la funció merge per a que utilitzi funcions lambda

def merge(x:list[any],y:list[any],key:any)->list[any]:  # Hem afegit un paràmetre key
    """
    Fusiona dues llistes prèviament ordenades en una nova llista ordenada
    :param
    x: una llista ordenada
    y: una llista ordenada
    :return
    una nova llista amb tots els elements d'x i y, ordenats
    """
    # print("MERGE: m'han cridat amb",x,y)  # xivato
    if len(x) < 1:
        return y
    if len(y) < 1:
        return x
    llista_nova = []
    
    i = j = 0
    while i < len(x) and j < len(y):
        if key(x[i]) <= key(y[j]):     # Afegeix el codi que falta
            llista_nova.append(x[i])
            i += 1
        else:
            llista_nova.append(y[j])
            j += 1
    llista_nova.extend(x[i:])
    llista_nova.extend(y[j:])
    return llista_nova
```

```python
# Codi adaptat de la funció mergesort per a que utilitzi funcions lambda

def mergesort(l:list[any], key: any)->list[any]:   # Afegeix el codi que falta
    """
    Ordena una llista segons la funció key
    :param
    l1: una llista a ordenar
    :return
    una nova llista ordenada
    """
    # print("MERGESORT: m'han cridat amb",l)  # xivato
    if len(l) < 2:
        return l
    else:
        meitat = len(l) // 2
        esquerra = mergesort(l[:meitat], key)  # Afegeix el codi que falta
        dreta = mergesort(l[meitat:], key)     # Afegeix el codi que falta
        return merge(esquerra, dreta, key)     # Afegeix el codi que falta
```

Farem una primera prova en la que la funció key serà el tipus de petició.

```python
key_tipus=lambda p: p["tipus"]

ordenades_tipus = mergesort(peticions, key_tipus)
print(ordenades_tipus[:5])
```

## ✍️ Exercici 3: Mostrar resultats

Escriu una funció que donada una llista de queixis la mostri per pantalla de manera ordenada i clara.
La funció ha de permetre indicar quants ítems mostrar del principi i quants del final.

Utilitza la plantilla donada.

```python
from typing import Any
```

```python

def mostra_queixes(queixes:list[any], n_inici:int, n_final:int)->None:
    """
    Mostra per pantalla una llista de queixes de manera ordenada i clara.
    :param
    queixes: llista de queixes a mostrar
    n_inici: nombre d'elements a mostrar del principi
    n_final: nombre d'elements a mostrar del final
    :return
    None
    """

    total_queixes: int = len(queixes)
    print(f"Mostrant {n_inici} elements del principi i {n_final} del final d'un total de {total_queixes} queixes:")

    for i in range(n_inici):
        print()
        print(i+1, end=". ")
        output_queixa(queixes[i])

    print()
    print("...")

    start_pos = total_queixes - n_final
    for i in range(n_final):
        curr_pos: int = start_pos + i
        print()
        print(curr_pos + 1, end=". ")
        output_queixa(queixes[curr_pos])
```

```python
def output_queixa(queixa: dict[str, Any]) -> None:
    """
    Given a queixa, print the queixa in format
    :param
    queixa: dict to be printed
    """
    format_dict = lambda key: key[0].upper() + key[1:] + ": " + str(queixa[key])  # for key of dict
    format_date = lambda q: (  # for queixa in queixes, output the expect format of DATA
        "Data: " + 
        str(q["dia"]) + "/" +
        str(q["mes"]) + "/" +
        str(q["any"])
    )

    print(format_dict("id") + ", " + format_dict("tipus") + ", " + format_dict("area"))
    print(format_date(queixa))
    print(format_dict("districte") + ", " + format_dict("barri"))
    print(format_dict("detall"))
```
