### 5.6 Negatius al davant

✍️ Escriu una funcio no recursiva anomenada negatius tal que, donada una llista de nombres enters, modifiqui la llista per col·locar els nombres negatius al principi de la llista. 

Assegura't que l'algorisme té complexitat lineal.

⚠️ No es poden fer servir funcions de Python que ajudin a resoldre el quid del problema tals i com min, max, pop, insert, sort, o altres de similars.

```python
def negatius(llista:list[int]):
    """
    Aquesta funció retorna la llista amb els nombres negatius al davant
    
    Parameters
    ----------
    llista: list
    
    Returns
    -------
    * no retorna res, però la llista inicial acaba amb els negatius al davant.
    """

    # i, j both points to the beginning
    # for i ++
    # if i <= 0
    #   i, j = j, i
    #   j++
    # i++

    i = j = 0;
    for i in range(len(llista)):
        if llista[i] < 0:
            llista[i], llista[j] = llista[j], llista[i]
            j += 1

    return llista
```

```python
# caso público
llista = [1, -2, 3, -4, -3, 5, 6]
negatius(llista)
```

```python
# todo positivo
llista = [1, 1, 2, 2, 3]
print(negatius(llista))
```

```python
# el primero negativo
llista = [-1, 3, 3, 4, 4]
print(negatius(llista))
```
