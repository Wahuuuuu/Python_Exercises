### 5.10 Comptador d'inversions en una llista

Volem ordenar una llista A de nombres enters en ordre creixent, es a dir,
$A[i] < A[j]$ per a qualsevol $i < j$.

Definim inversió, com el procés d'intercanviar dos valors de la llista que no
estiguin en el seu ordre natural, es a dir, que el de l'esquerra sigui més gran
que el de la dreta. Concretament, diem que $(A[i];A[j])$ es una inversió, si
$A[i] > A[j]$ i $i < j$. 

Per exemple:
\
Si es té una llista **[1, 3, 2, 4]**, aleshores hi hauria una inversió **(3,2)**, ja que **3 > 2** i l'index de la posició del 3 és més petit que la del 2. En altres paraules, es troba el 3 més a l'esquerra de la llista que el 2. Per tant, només es produiria 1 inversió.

### ✍️ Escriu una funció que calculi el nombre d'inversions dins una llista i que digui quines son.

```python
"""
    Author: Yanzhe Chen
    Date: 26/11/2025
    Version: 1.0
"""

def comptador_inversions(llista: list[int]) -> int:
    """
    Aquesta funció compta quantes inversions s'han fet.
    
    Parameters
    ----------
    llista: list
    
    Returns
    -------
    num: int
    """
    
    num_inversions: int = merge_count(llista, 0, len(llista) - 1)

    return num_inversions


def merge_count(llista: list[int], left: int, right: int) -> int:
    if left >= right:
        return 0
    
    mid: int = (left + right) // 2
    
    left_num: int = merge_count(llista, left, mid)
    right_num: int = merge_count(llista, mid+1, right)

    curr_num: int = sort_and_count(llista, left, mid, right)

    return left_num + right_num + curr_num


def sort_and_count(llista: list[int], left: int, mid: int, right: int) -> int:
    """
    [left, mid] and [mid+1, right] are sorted increasingly
    left <= mid < right

    This function returns the number of inversion pairs between [left, mid] and [mid+1, right]
    [left, right] will be sorted once the function ends
    """

    sorted: list[int] = []
    count: int = 0

    # known that left <= i <= mid and mid+1 <= j <= right
    # if llista[i] is bigger than llista[j],
    # means that every single element in [i : mid+1] is an inversion pair with llista[j]
    # there are (mid + 1) - i inversion pairs
    i: int = left; j: int = mid + 1
    while (i <= mid and j <= right):
        if llista[i] > llista[j]:
            count += (mid + 1 - i)
            sorted.append(llista[j])
            j += 1
        else:
            sorted.append(llista[i])
            i += 1
    
    if i <= mid:
        sorted.extend(llista[i : mid+1])
    if j <= right:
        sorted.extend(llista[j : right+1])
    
    # fill the sorted list
    for p in range(0, len(sorted)):
        llista[left + p] = sorted[p]
    
    return count
```
